```python
import graphviz
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from copy import deepcopy

mpl.rcParams['xtick.major.size'] = 0
mpl.rcParams['ytick.major.size'] = 0
mpl.rcParams['axes.labelsize'] = 12
mpl.rcParams['grid.alpha'] = 0.5

mpl.rcParams['font.family'] = 'sans-serif'
mpl.rcParams['font.sans-serif'] = ['Helvetica Neue']

mpl.rcParams['savefig.bbox'] = 'tight'
mpl.rcParams['savefig.format'] = 'pdf'
```

## Model schematic

```python
g = graphviz.Digraph(filename='model', strict=True, directory='../Figures', format='svg')

g.node('Blood infection', style='filled', color='DarkRed', fillcolor='DarkRed', fontname='Helvetica', fontcolor='White')
g.node('Infectious bite', fontname='Helvetica')
g.node('Liver infection', fontname='Helvetica')
g.edge('Infectious bite', 'Liver infection', penwidth='8')
g.edge('Liver infection', 'Blood infection', penwidth='8')

g.node('Clinical', style='filled', color='DarkRed', fillcolor='DarkRed', fontcolor='White', fontname="Helvetica")
g.edge('Blood infection', 'Clinical', penwidth='7', weight='2')

g.node('Severe', style='filled', color='DarkRed', fillcolor='DarkRed', fontcolor='White', fontname='Helvetica')
g.edge('Clinical', 'Severe', penwidth='2', weight='2')

g.node('Death', style='filled', color='DarkRed', fillcolor='DarkRed', fontcolor='White', fontname='Helvetica')
g.edge('Severe', 'Death', weight='2', penwidth='1')

g.render()
g
```

## Clinical and SM immunity

```python
g2 = deepcopy(g)
g2.filename = 'natural_immunity'


g2.edge('Blood infection', 'Clinical', label='  clinical immunity (slow)', arrowhead='lnormal', fontname='Helvetica', fontcolor='Green', color='Green')
g2.edge('Clinical', 'Severe', label='  severe malaria immunity (fast)', arrowhead='lnormal', fontname='Helvetica', fontcolor='Green', color='Green')

g2.render()
g2
```

## Liver stage vaccine

```python
g3 = deepcopy(g)
g3.filename = 'liver_stage_vac'

g3.edge('Infectious bite', 'Liver infection', '   pre-erythrocytic vaccine?', arrowhead='rnormal', fontname='Helvetica', fontcolor='Orange', color='Orange')
g3.edge('Liver infection', 'Blood infection', '   pre-erythrocytic vaccine?', arrowhead='rnormal', fontname='Helvetica', fontcolor='Orange', color='Orange')

g3.render()
g3
```

### Blood stage vaccine

```python
g4 = deepcopy(g)
g4.filename = 'blood_stage_vac'

g4.edge('Liver infection', 'Blood infection', '   blood stage vaccine?', arrowhead='rnormal', fontname='Helvetica', fontcolor='Orange', color='Orange')
g4.edge('Blood infection', 'Clinical', '   blood stage vaccine?', arrowhead='rnormal', fontname='Helvetica', fontcolor='Orange', color='Orange')
g4.edge('Clinical', 'Severe', '   blood stage vaccine?', arrowhead='rnormal', fontname='Helvetica', fontcolor='Orange', color='Orange')


g4.render()
g4
```
