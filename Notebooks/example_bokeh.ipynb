```python
# model is declared in plot_bokeh as it's not directly passed to plot_bokeh.plot()
# set which variables to display
model.variables = Variables(['All infection', 'All clinical', 'Severe malaria', 'Direct deaths'])

# set which measures of each variable to display
model.measures = Measures(['cases', 'cdf', 'averted', 'efficacy'])

# set plot configuration parameters
model.Config(time_scale='Years', fig_xscale=0.75, fig_yscale=0.6, show_season=True, show_vac=True, show_smc=True, show_death_rates=True)

# pass simulation function and panels function
model.Sources(simulation_fn=simulate_single_vaccine, panels_fn=panels)

# set simulation model parameters, see ../Edinburgh_Model/model.py for parameter definitions
model.pars = Parameters(
    lsv=True,
    bsv=False,
    smc=False,
    popsize=100000,
    λ=1,
    β=0,
    children='cohort',
    study_months=21*12,
    season_peak=0,
    season_width=0.1,
    min_vac_age=int(5*weeks_per_month),
    vac_age_range=int(12*weeks_per_month),
    smc_offset=0,
    smc_repeats=4,
    nboosters_blood=1,
    nboosters_liver=1,
    protection_blood=False,
    recording='Last primary dose',
    control='Control',
    treatment='R21',
)

# initialise sliders and buttons (do after initialising parameters)
model.Sliders('all')
model.Buttons('all')

show(plot, notebook_url="http://localhost:8888")
```
